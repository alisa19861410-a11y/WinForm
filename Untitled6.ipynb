```python
import numpy as np
import time

# строки таблицы 3
T = np.array([
 [13.5,384,2.1,1.54,190,.173,-.697],
 [36.1,380,1.1,1.19,231,.067,-.499],
 [38.6,68,4.7,.90,110,.078,.152],
 [38.4,154,10.6,.88,131,.042,.550],
 [24.2,382,6.2,1.49,223,.088,-.335],
 [27.1,277,4.5,1.27,119,.090,1.147],
 [17.6,267,12.0,.80,120,.088,.502],
 [25.5,103,8.6,1.17,144,.080,-.373],
 [18.0,322,9.5,.52,140,.041,-.054],
 [24.1,285,6.6,1.88,186,.060,.439],
 [18.1,128,11.1,1.14,141,.050,-.486],
 [13.7,341,9.1,.88,164,.143,-.347]
], dtype=float)

d, h, a, speed, cost, popularity, R0 = T.T
K, n = 12, 48

# вектор максимально допустимых правок для каждого параметра
delta = np.tile([.20, .20, .20, .10], K)
# коэффициенты чувствительности для каждого типа параметра
theta = np.array([1., 1., .4, .8])
# Итоговые веса заметности
omega = (popularity[:, None] * theta).ravel()
# матрица G показывает как правки параметров влияют на боевой рейтинг каждого юнита
G = np.zeros((K, n))
for i in range(K):
    G[i, 4*i:4*i+4] = [
        1.5*speed[i]*d[i]/27.825,
        1.5*h[i]/257.56,
        .5*a[i]/7.163,
        -3.5*cost[i]/158.19
    ]
# матрица A ограничения на сохранение средних значений по ростеру
A = np.zeros((4, n))
for j, values in enumerate([d, h, a, cost]):
    A[j, j::4] = values
A /= np.linalg.norm(A, axis=1)[:, None]

pairs = [(i,j) for i in range(K) for j in range(i+1,K)]

# матрица B вычисляет разности рейтингов для каждой пары
B = np.zeros((len(pairs), K))
for s,(i,j) in enumerate(pairs):
    B[s,i], B[s,j] = 1., -1.

# веса пар зависящие от популярности обоих юнитов в паре
W = np.array([popularity[i]*popularity[j] for i,j in pairs])
W /= W.sum()   # нормировка весов чтобы их сумма была равна 1

# ширина коридора и порог для разности рейтингов
eps = .05
window = np.log((.5+eps)/(.5-eps))

# функция валидации принимает вектор правок x, считает новые рейтинги, винрейты и проверяет выполнение всех ограничений, возвращает словарь с метриками
def check(x):
    ratings = R0 + G @ x   # новые рейтинги
    diff = B @ ratings   # разности рейтингов
    winrate = 1. / (1. + np.exp(-diff))   # винрейты через логистическую функцию

    return dict(
        winrate_min=float(winrate.min()),
        winrate_max=float(winrate.max()),
        # Количество пар, которые вышли за пределы коридора
        pairs_outside=int(np.sum((winrate < .45-1e-8) | (winrate > .55+1e-8))),
        equality_error=float(np.max(np.abs(A @ x))),    # ошибка выполнения Ax=0
        interval_violation=float(np.max(np.maximum(np.abs(x)-delta,0))), # выход за интервалы
        max_interval_ratio=float(np.max(np.abs(x)/delta)),     # отношение макс правки к лимиту
        weighted_L1=float(omega @ np.abs(x)),     # взвешенная L1 норма
        changed_components=int(np.sum(np.abs(x)>1e-7)),     # количество измененных параметров
        rating_spread=float(np.ptp(ratings)),     # разброс рейтингов
        ellipsoid_radius=float(np.linalg.norm(x/delta))    # норма в метрике эллипсоида
    )

# функция сохранения
def finish(method_name, name, x, details):
    # запись патча в csv
    np.savetxt(
        name+'_patch.csv', x.reshape(K,4), delimiter=',',
        header='x_damage,x_health,x_armor,x_cost',
        comments=''
    )
    print(f" {method_name.upper()}")
    print("\n ДЕТАЛИ ВЫЧИСЛЕНИЙ ")
    # Словарь для перевода ключей на русский
    ru_details = {
        'iterations': 'Итераций',
        'seconds': 'Время (сек)',
        't_star': 'Разброс рейтингов (t*)',
        'epsilon_star': 'Мин. коридор (eps*)'
    }
    for key, val in details.items():
        ru_key = ru_details.get(key, key)
        if isinstance(val, float):
            print(f"  {ru_key:25}: {val:.6f}")
        else:
            print(f"  {ru_key:25}: {val}")

    val_dict = check(x)
    ru_val = {
        'winrate_min': 'Мин. винрейт',
        'winrate_max': 'Макс. винрейт',
        'pairs_outside': 'Пар вне коридора',
        'equality_error': 'Ошибка Ax=0',
        'interval_violation': 'Выход за интервалы',
        'max_interval_ratio': 'Макс. отношение к лимиту',
        'weighted_L1': 'Взвешенная L1-норма',
        'changed_components': 'Изменено компонент',
        'rating_spread': 'Разброс рейтингов',
        'ellipsoid_radius': 'Радиус эллипсоида'
    }
    print("\n ПРОВЕРКА ОГРАНИЧЕНИЙ ")
    for key, val in val_dict.items():
        ru_key = ru_val.get(key, key)
        if isinstance(val, float):
            print(f"  {ru_key:25}: {val:.6e}")
        else:
            print(f"  {ru_key:25}: {val}")

def simplex(c, E, f, U, v, tol=1e-9, cap=30000):
    n0 = len(c)      # количество исходных переменных
    ne, nu = len(f), len(v) # количество ограничений-равенств и неравенств
    rows = []     # строки симплекс-таблицы
    rhs = []    # правые части ограничений
    basic = []   # индексы базисных переменных
    art = []   # индексы искусственных переменных
    size = n0   # текущий размер задачи

    # построение начальной таблицы
    for i in range(ne+nu):
        a0 = E[i] if i < ne else U[i-ne]
        bi = float(f[i] if i < ne else v[i-ne])
        sign = 1 if bi >= 0 else -1
        row = {
            j:float(t*sign)
            for j,t in enumerate(a0)
            if abs(t)>1e-14
        }

        # для ограничений-неравенств добавляем балансовую переменную
        if i >= ne:
            row[size] = float(sign)
            slack = size
            size += 1
        # для = и для => добавляем искусственную переменную
        if i < ne or sign < 0:
            row[size] = 1.
            art.append(size)
            basic.append(size)
            size += 1
        else:
            basic.append(slack)
        rows.append(row)
        rhs.append(abs(bi))

    m = len(rows)
    tab = np.zeros((m+1,size+1)) # инициализация симплекс-таблицы
    # заполняем таблицу
    for i,row in enumerate(rows):
        for j,value in row.items():
            tab[i,j] = value
        tab[i,-1] = rhs[i]
    count = 0
# пересчет целевой строки на основе текущего базиса
    def objective(cost):
        tab[-1,:-1] = cost
        tab[-1,-1] = 0.
        for i,j in enumerate(basic):
            tab[-1,:] -= cost[j] * tab[i,:]
# поиск разрешающего элемента и пересчет таблицы
    def pivots(allowed):
        nonlocal count

        while count < cap:
            # ищем небазисные столбцы с отрицательной оценкой
            enter = np.flatnonzero(
                (tab[-1,:-1] < -tol) & allowed
            )
            if not len(enter):
                return

            j = int(enter[0])
            # ищем строки где коэффициент положителен
            possible = np.flatnonzero(
                tab[:len(basic),j] > tol
            )
            if not len(possible):
                raise RuntimeError(
                    'Целевая функция не ограничена'
                )

            # правило минимального отношения для выбора ведущей строки
            ratios = tab[possible,-1] / tab[possible,j]
            rmin = float(np.min(ratios))
            tied = possible[ratios <= rmin+1e-9]
            i = int(min(tied,key=lambda z:basic[z])) # правило бленда против зацикливания
            # гауссово исключение
            tab[i,:] /= tab[i,j]
            factors = tab[:,j].copy()
            factors[i] = 0.
            tab[:] -= factors[:,None] * tab[i:i+1,:]
            basic[i] = j
            count += 1
        raise RuntimeError('Превышен лимит итераций')

    # поиск начального допустимого базиса
    cost1 = np.zeros(size)
    cost1[art] = 1. # минимизируем сумму искусственных переменных
    objective(cost1)
    pivots(np.ones(size,dtype=bool))
    # если минимум не равен нулю значит допустимого решения не существует
    if -tab[-1,-1] > 1e-6:
        raise RuntimeError('Нет допустимого решения')
    # удаление искусственных переменных из базиса
    aset = set(art)
    redundant = []

    for i in range(len(basic)):
        if basic[i] in aset:
            # ищем неискусственную переменную для замены
            choices = [
                j for j in range(size)
                if j not in aset
                and j not in basic
                and abs(tab[i,j])>tol
            ]
            if choices:
                j = choices[0]
                tab[i,:] /= tab[i,j]
                factors = tab[:,j].copy()
                factors[i] = 0.
                tab[:] -= factors[:,None] * tab[i:i+1,:]
                basic[i] = j
            else:
                if abs(tab[i,-1])>1e-6:
                    raise RuntimeError(
                        'Нет допустимого решения'
                    )
                redundant.append(i)
    # удаляем пустые строки
    if redundant:
        mask = [
            i for i in range(len(basic))
            if i not in redundant
        ]
        tab = tab[mask+[-1],:]
        basic = [basic[i] for i in mask]
    # переиндексация столбцов после удаления искусственных переменных
    columns = [
        j for j in range(size)
        if j not in aset
    ]
    renumber = {
        j:i for i,j in enumerate(columns)
    }
    tab = np.column_stack([
        tab[:,columns],tab[:,-1]
    ])
    basic = [renumber[j] for j in basic]

    # минимизация реальной целевой функции
    cost2 = np.zeros(len(columns))
    cost2[:n0] = c
    objective(cost2)
    pivots(np.ones(len(columns),dtype=bool))
    # изначений исходных переменных
    z = np.zeros(n0)
    for i,j in enumerate(basic):
        if j<n0:
            z[j] = tab[i,-1]
    return z,count

start = time.perf_counter() # засекаем время

# переменные y = x + delta => 0, v = v_plus - v_minus, t => 0
c = np.zeros(n+3)
c[-1] = 1. # минимизируем t разброс рейтингов
# ограничения-равенства Ax = 0  =>  A(y - delta) = 0  =>  Ay = Adelta
E = np.zeros((4,n+3))
E[:,:n] = A
f = A @ delta
# ограничения-неравенства
U = []
v = []

for i in range(K):
    # ограничение R_i(x) - v <= 0 =>  G_iy - v_plus + v_minus - R0_i - G_idelta <= 0
    row = np.zeros(n+3)
    row[:n] = G[i]
    row[n] = -1.
    row[n+1] = 1.
    row[-1] = -1.
    U.append(row)
    v.append(-R0[i]+G[i]@delta)
    # ограничение -R_i(x) + v <= 0
    row = np.zeros(n+3)
    row[:n] = -G[i]
    row[n] = 1.
    row[n+1] = -1.
    U.append(row)
    v.append(R0[i]-G[i]@delta)
for j in range(n):
    # ограничение: y_j <= 2delta_j (верхняя граница для сдвинутых переменных)
    row = np.zeros(n+3)
    row[j] = 1.
    U.append(row)
    v.append(2*delta[j])

# запуск симплекс-метода
z,iterations = simplex(
    c,E,f,np.array(U),np.array(v)
)
# возврат к исходным переменным x
x = z[:n] - delta

assert max(abs(A@x))<1e-6              # ср знач сохр
assert max(abs(x)-delta)<1e-6           # правки не выходят за границы
assert np.ptp(R0+G@x)<=z[-1]+1e-6       # разброс рейтингов не превышает t*

elapsed = time.perf_counter()-start # останавливаем время

finish('Симплекс-метод', 'method1', x, dict(
    iterations=iterations,                 # количество итераций симплекса
    seconds=elapsed,                    # время выполнения
    t_star=float(z[-1]),                  # минимальный разброс рейтингов
    epsilon_star=float(
        1/(1+np.exp(-z[-1]))-.5          # минимально достижимый коридор
    )
))
```
